areas, such as cleaner air and water and flood mitigation. However, the

structure of greenspace and human health. These types of metrics include
mean patch size, edge density, and fragmentation.

### Download census tracts and select your urban area

```python
# Download the census tracts for state of IL (only once) Part 1 of 1

# Define info for census tract download
tiger_tracts_dir = os.path.join(census_tracts_data_dir, 'illionis-tracts')
os.makedirs(tiger_tracts_dir, exist_ok=True)
tiger_tracts_path = os.path.join(tiger_tracts_dir, '*.shp')

# Only download once (conditional statement)
if not os.path.exists(tiger_tracts_path):
    tiger_tracts_url = (
    'https://www2.census.gov/geo/tiger/TIGER2024/TRACT/tl_2024_17_tract.zip'
    )
    tiger_tracts_gdf = gpd.read_file(tiger_tracts_url)
    # COUNTYFP 031 is Cook County which contains the city of Chicago
    chi_tiger_tracts_gdf = tiger_tracts_gdf[tiger_tracts_gdf.COUNTYFP=='031']
    chi_tiger_tracts_gdf.to_file(tiger_tracts_path, index=False)

# Load in the census tract data
chi_tiger_tracts_gdf = gpd.read_file(tiger_tracts_path)

# Call the chicago tracts gdf to see it
chi_tiger_tracts_gdf.head()
```

```python
# Perform a spatial join for census tracts at least partially 
# within City of Chicago boundary

# this new gdf needs to be joined to the previous one from CDC which
# is already clipped to the city boundary, so no need to download a 
# seperate city boundary shapefile which reduces the amount of things 
# being downloaded

# Define new variable for the joind gdf
joined_chi_tracts_gdf = (
    gpd.sjoin(
        # TIGER tracts gdf - only need tracts that intersect with..
        chi_tiger_tracts_gdf.to_crs(ccrs.Mercator()),
        # CDC tracts gdf - which are already clipped to the Chicago city boundary
        chicago_tracts_gdf.to_crs(ccrs.Mercator()), 
        # Specify type of join ("inner", "left", "right")
        how="inner", 
        # Specify the spatial relationship ("intersects", "within", "contains")
        predicate="intersects"
        )
)

# Explore the result
joined_chi_tracts_gdf.head()
```

```python
# Try to see how many rows there are because I think there's duplicates
num_rows = joined_chi_tracts_gdf.shape[0]
print("Number of rows:", num_rows)
```

```python
# Drop duplicate geometries

# Normalize the geometry column to ensure consistent representation
joined_chi_tracts_gdf['geometry'] = joined_chi_tracts_gdf.geometry.normalize()

# Drop duplicate rows based on the geometry column
dropped_joined_chi_tracts_gdf = joined_chi_tracts_gdf.drop_duplicates(subset='geometry')

dropped_joined_chi_tracts_gdf.head()
```

```python
# Site plot -- Census tracts with satellite imagery in the background
# Create new variable for plot in order to save it later
joined_chicago_tracts_plot = dropped_joined_chi_tracts_gdf.to_crs(
# Use hvplot to plot and set parameters
ccrs.Mercator()).hvplot(
    geo=True, crs=ccrs.Mercator(),
    tiles='EsriImagery',
    title='City of Chicago - Site Plot',
    fill_color=None, line_color='darkorange', 
    line_width=3, frame_width=600,
    xaxis=None, yaxis=None,
)

# Save the plot as html to be able to display online
hv.save(joined_chicago_tracts_plot, 'joined_chi_site_plot_using_tiger_and_cdc.html')  

# Display the plot
joined_chicago_tracts_plot
```

```python
# Set up a path for the asthma data
cdc_asthma_path = os.path.join(census_tracts_data_dir, 'asthma.csv')

# Download asthma data (only once)
if not os.path.exists(cdc_asthma_path):
    # Define new variable for url
    cdc_asthma_cook_county_url = (
        "https://data.cdc.gov/resource/cwsq-ngmh.csv"
        "?year=2022"
        "&stateabbr=IL"
        "&countyname=Cook"
        "&measureid=CASTHMA"
        "&$limit=1500"
    )
    # Define new variable for dataframe
    cdc_asthma_df = (
        # Read a CSV file into a dataframe
        pd.read_csv(cdc_asthma_cook_county_url)
        # Replace column names as needed - 'old_name':'new_name'
        .rename(columns={
            'data_value': 'asthma',
            'low_confidence_limit': 'asthma_ci_low',
            'high_confidence_limit': 'asthma_ci_high',
            'locationname': 'tract'})
        # Select specifc columns needed/wanted with double brackets    
        [[
            'year', 'tract', 
            'asthma', 'asthma_ci_low', 'asthma_ci_high', 'data_value_unit',
            'totalpopulation', 'totalpop18plus'
        ]]
    )
    # Save dataframe to a CSV (tabular data) file
    cdc_asthma_df.to_csv(
        cdc_asthma_path, 
        # Prevent a new index column from being created
        index=False
        )

# Load in asthma data
cdc_asthma_df = pd.read_csv(cdc_asthma_path)

# Preview asthma data
cdc_asthma_df.head()
```

### Join health data with census tract boundaries

```python
# Change tract identifier datatype for merging
dropped_joined_chi_tracts_gdf.tract2010 = dropped_joined_chi_tracts_gdf.tract2010.astype('int64')

# Merge census data with geometry
tract_cdc_gdf = (
    # Census tracts gdf
    dropped_joined_chi_tracts_gdf
    # Use the .merge() method
    .merge(
        # Asthma prevalence dataframe
        cdc_asthma_df, 
        # Specify the column/index to merge on for Census tracts gdf
        left_on='tract2010', 
        # Specify the column/index to merge on for the asthma dataframe
        right_on='tract', 
        # Specify type of join ("inner", "left", "right")
        how='inner'
        )
)

# Plot asthma data as chloropleth
chi_chloropleth_asthma_by_tract = (
# Use EsriImagery tiles for the background
(   gv.tile_sources.EsriImagery
    * 
    gv.Polygons(
        # Change gdf CRS to mercator
        tract_cdc_gdf.to_crs(ccrs.Mercator()),
        # Set variables for the plot
        vdims=['asthma', 'tract2010'],
        # Set CRS to Mercator
        crs=ccrs.Mercator()
    ).opts(
        # Add a colorbar and label
        color='asthma', colorbar=True, 
        clabel='% of Adult Population',
        tools=['hover'])
).opts(
    # Plot size, title, and axes labels
    title= 'City of Chicago - Asthma Prevalence by Census Tract',
    width=600, height=600, 
    xaxis=None, yaxis=None,
    )
)

# Save the plot as html to be able to display online
hv.save(chi_chloropleth_asthma_by_tract, 'chi_chloropleth_asthma_by_tract.html')  

# Display the plot
chi_chloropleth_asthma_by_tract
```

### Compute NDVI Statistics

Next, calculate some metrics to get at different aspects of the
distribution of greenspace in each census tract. These types of
statistics are called **fragmentation statistics**, and can be
implemented with the `scipy` package. Some examples of fragmentation
statistics are:

Percentage vegetation  
The percentage of pixels that exceed a vegetation threshold (.12 is
common with Landsat) Mean patch size

The average size of **patches**, or contiguous area exceeding the
vegetation threshold. Patches can be identified with the `label`
function from `scipy.ndimage` Edge density

The proportion of edge pixels among vegetated pixels. Edges can be
identified by **convolving** the image with a **kernel** designed to
detect pixels that are different from their surroundings.

> **What is convolution?**
>
> If you are familiar with differential equations, convolution is an
> approximation of the LaPlace transform.
>
> For the purposes of calculating edge density, convolution means that
> we are taking all the possible 3x3 chunks for our image, and
> multiplying it by the kernel:
>
> $$
> \text{Kernel} =
> \begin{bmatrix}
> 1 & 1 & 1 \\
> 1 & -8 & 1 \\
> 1 & 1 & 1
> \end{bmatrix}
> $$
>
> The result is a matrix the same size as the original, minus the
> outermost edge. If the center pixel is the same as the surroundings,
> its value in the final matrix will be
> $-8 + 1 + 1 + 1 + 1 + 1 + 1 + 1 + 1 = 0$. If it is higher than the
> surroundings, the result will be negative, and if it is lower than the
> surroundings, the result will be positive. As such, the edge pixels of
> our patches will be negative.

```python
# Skip this step if data are already downloaded 
if not chi_scene_df is None:
    # Get an example tract
    ex_tract = dropped_joined_chi_tracts_gdf.loc[0].tract2010
    # Get example gdf based off of the example tract
    chi_ex_scene_gdf = chi_scene_df[chi_scene_df.tract==ex_tract]

    # Loop through all images for tract
    tile_das = []
    # Create for loop, iterate over rows
    for _, href_s in chi_ex_scene_gdf.iterrows():
        # Open vsi connection to data
        tile_da = rxr.open_rasterio(
            # File path/url to the multispectral image
            href_s.rgbir_href, 
            # Create masked array, then remove any single-dimensional axes from it
            masked=True).squeeze()
        
        # Crop data to the bounding box of the census tract
        # Create the boundary
        boundary = (
            # Using the census tract gdf
            tract_cdc_gdf
            # Set the 'tract2010' as the index of the gdf
            .set_index('tract2010')
            # Select the ex_tract from that gdf
            .loc[[ex_tract]]
            # Set to the same CRS as the images for the tract
            .to_crs(tile_da.rio.crs)
            # Access the geometry of this tract to perform further operations on it
            .geometry
        )
        # Crop the data to bounding box
        crop_da = tile_da.rio.clip_box(
            # Compute bounding box (min and max coordinates) of census tract geometry
            *boundary.envelope.total_bounds,
            # Expand bounding box slightly beyond its original extent to ensure full coverage
            auto_expand=True)

        # Clip data to the boundary of the census tract
        clip_da = crop_da.rio.clip(boundary, all_touched=True)

        # Compute NDVI ((NIR - Red)/(NIR + Red))
        ndvi_da = (
            (clip_da.sel(band=4) - clip_da.sel(band=1)) 
            / (clip_da.sel(band=4) + clip_da.sel(band=1))
        )

        # Accumulate result
        tile_das.append(ndvi_da)

    # Merge data
    scene_da = merge_arrays(tile_das)

    # Mask vegetation
    veg_mask = (scene_da>.3)

    # Calculate mean patch size
    # Label the connected areas
    labeled_patches, num_patches = label(veg_mask)
    # Count number of non-negative occurances, flatten the array to 1D
    patch_sizes = np.bincount(labeled_patches.ravel())[1:]
    # Get the mean 
    mean_patch_size = patch_sizes.mean()

    # Calculate edge density
    kernel = np.array([
        [1, 1, 1], 
        [1, -8, 1], 
        [1, 1, 1]])
    # Apply convolution to the vegetation mask
    edges = convolve(
        # Input array - the array to apply convolution on
        veg_mask, 
        # Kernel array - the smaller array that defines the filter to be applied
        kernel, 
        # Input array is extended beyond its boundaries by - 
        # filling all values beyond the edge with the same constant value
        mode='constant')
    
    # Calculate edge density = 
    # count of edge pixels present / total number of pixels in the veg_mask
    edge_density = np.sum(edges != 0) / veg_mask.size
```

### Spatial bias

We always need to think about bias, or systematic error, in model
results. Every model is going to have some error, but we’d like to see
that error evenly distributed. When the error is systematic, it can be
an indication that we are missing something important in the model.

In geographic data, it is common for location to be a factor that
doesn’t get incorporated into models. After all – we generally expect
places that are right next to each other to be more similar than places
that are far away (this phenomenon is known as *spatial
autocorrelation*). However, models like this linear regression don’t
take location into account at all.
